# Day 3: Linear Algebra with NumPy

Vectors, matrices, dot products and matrix multiplication, then a look at why they matter:
a linear model's prediction is a dot product.

```python
import numpy as np
```

## Vectors

```python
a = np.array([2, 3])
b = np.array([4, 1])

print("a + b  :", a + b)
print("a * 2  :", a * 2)
print("shape  :", a.shape)

v = np.array([3, 4])
print("magnitude of [3, 4]:", np.linalg.norm(v))
```

## Dot product: multiply pairs, then add

```python
manual = sum(x * y for x, y in zip(a, b))
print("manual  :", manual)
print("np.dot  :", np.dot(a, b))
print("a @ b   :", a @ b)
```

## Matrices and matrix multiplication

```python
A = np.array([[1, 2],
              [3, 4]])
B = np.array([[5, 6],
              [7, 8]])

print("A @ B (matrix multiplication):\n", A @ B)
print("\nA * B (element by element, different thing!):\n", A * B)
print("\nTranspose of A:\n", A.T)
```

```python
# Shape rule: (m x n) @ (n x p) -> (m x p)
C = np.ones((3, 2))
D = np.ones((2, 4))
print("(3x2) @ (2x4) ->", (C @ D).shape)

try:
    D @ C
except ValueError as e:
    print("\n(2x4) @ (3x2) fails:", e)
```

## Why it matters: a dataset is a matrix
Rows are samples, columns are features. A linear model predicts with one matrix multiplication:
`predictions = X @ w + b`.

```python
X = np.array([[5, 2],     # house 1: area, bedrooms
              [10, 4],    # house 2
              [7, 3]])    # house 3
w = np.array([0.5, 0.8])  # weights: price per marla, price per bedroom
b0 = 1.0                  # bias (intercept)

print("X shape:", X.shape)
print("Predictions:", X @ w + b0)

new_house = np.array([5, 3])
print("5 marla, 3 bedrooms ->", new_house @ w + b0, "million PKR")
```

## Bonus: linear regression with just linear algebra
Fit `y = b0 + b1 * x` using the normal equation `(X^T X) beta = X^T y`. This is the same answer
Scikit-learn will give on Day 5.

```python
hours = np.array([1, 2, 3, 4, 5])
marks = np.array([2, 4, 5, 4, 5])

X_design = np.column_stack([np.ones_like(hours), hours])   # first column of ones is the intercept
beta = np.linalg.solve(X_design.T @ X_design, X_design.T @ marks)

print("intercept b0 =", round(beta[0], 2))
print("slope     b1 =", round(beta[1], 2))
print("prediction for 6 hours:", round(beta[0] + beta[1] * 6, 2))
```
