# Day 1: Machine Learning Foundations

Goal for today: get comfortable with the vocabulary (sample, feature, target) by looking at a small
house dataset in Pandas, and tell supervised problems apart from the rest.

## AI, ML and DL in one line each
- **AI**: any system that does something we would call intelligent.
- **ML**: a part of AI where the computer learns patterns from data instead of hand-written rules.
- **DL**: a part of ML that uses neural networks with many layers.

```python
import pandas as pd

df = pd.read_csv("toy_houses.csv")
df
```

## Samples, features and target
Each row is one **sample** (one house). The input columns are the **features**. The column we want
to predict is the **target** (also called the label).

```python
target_col = "price_million_pkr"
feature_cols = ["area_marla", "bedrooms"]

X = df[feature_cols]   # features
y = df[target_col]     # target

print("Samples :", X.shape[0])
print("Features:", X.shape[1], feature_cols)
print("Target  :", target_col)
print("X shape :", X.shape, "| y shape:", y.shape)
```

## Regression or classification?
If the target is a number we are predicting a quantity (regression). If it is a category we are
predicting a class (classification).

```python
print("Target dtype:", y.dtype)
print("Problem type:", "regression" if pd.api.types.is_numeric_dtype(y) else "classification")

# A tiny classification example: spam or not spam
emails = pd.DataFrame({
    "has_word_lottery": [1, 0, 1, 0, 0, 1],
    "num_links":        [5, 1, 7, 0, 2, 6],
    "is_spam":          ["yes", "no", "yes", "no", "no", "yes"],
})
print(emails)
print("\nTarget dtype:", emails["is_spam"].dtype)
print("Classes:", emails["is_spam"].unique())
```

## Types of learning
| Type | Data | Example from this week |
|---|---|---|
| Supervised | labelled | house price (regression), high or low price (classification) |
| Unsupervised | unlabelled | grouping customers by behaviour (not covered this week) |
| Reinforcement | rewards from an environment | a game-playing agent (not covered this week) |

```python
tasks = {
    "Predict house price from area and bedrooms": "supervised (regression)",
    "Decide if an email is spam":                 "supervised (classification)",
    "Group customers with similar habits":        "unsupervised (clustering)",
    "Teach a robot to walk using rewards":        "reinforcement learning",
}
for task, kind in tasks.items():
    print(f"{task:45s} -> {kind}")
```
